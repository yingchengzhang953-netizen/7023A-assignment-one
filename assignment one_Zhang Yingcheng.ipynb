```python
#!/usr/bin/env python3
"""Super Tool: Calculator + QA Bot in a single file with a main menu."""

from decimal import Decimal, InvalidOperation

def main():
    print("====== Welcome! Please choose a tool ======")
    while True:
        print("\n========== Main Menu ==========")
        print("  1. Calculator")
        print("  2. QA Bot")
        print("  3. Exit")
        choice = input("Please choose (1/2/3): ").strip()

        if choice == "1":
            run_calculator()
        elif choice == "2":
            run_qa_bot()
        elif choice == "3":
            print("Goodbye! See you next time.")
            break
        else:
            print("Invalid choice. Please enter 1, 2 or 3.")


if __name__ == "__main__":
    main()
```

```python

# ============================================================
#  Part 1: Calculator
# ============================================================

def calc_read_number(prompt):
    """Keep asking until the user enters a valid number, one at a time."""
    while True:
        try:
            return Decimal(input(prompt).strip())
        except InvalidOperation:
            print("Invalid input. Please enter ONE number, e.g. 2.1")

def calc_read_operator():
    """Keep asking until the user enters one of +, -, *, /."""
    while True:
        op = input("Choose operator (+, -, *, /): ").strip()
        if op in ("+", "-", "*", "/"):
            return op
        print("Invalid operator. Please enter one of: +, -, *, /")

def calc_calculate(a, b, op):
    if op == "+":
        return a + b
    if op == "-":
        return a - b
    if op == "*":
        return a * b
    # op == "/"
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return a / b

def run_calculator():
    """Calculator mode."""
    print("\n=== Simple Calculator ===")
    print("Note: enter numbers ONE at a time (first number, then second number).")
    while True:
        a = calc_read_number("Enter the first number: ")
        b = calc_read_number("Enter the second number: ")
        op = calc_read_operator()
        try:
            result = calc_calculate(a, b, op)
        except ZeroDivisionError as e:
            print(f"Error: {e}")
        else:
            print(f"Result: {a} {op} {b} = {result}")

        again = input("Calculate again? (y/n): ").strip().lower()
        if again != "y":
            print("Goodbye!")
            break
```

```python

# ============================================================
#  Part 2: QA Bot
# ============================================================

BOT_NAME = "Clever"

def qa_answer(question):
    """Match the question against keywords and return an answer."""
    q = question.lower().strip()

    if "hello" in q or "hi" in q or "hey" in q:
        return "Hello there! Nice to meet you. Ask me anything."
    elif "python" in q:
        return "Python is a versatile, easy-to-learn programming language. " \
               "It's great for scripting, data science, web development, and AI."
    elif "jetson" in q:
        return "NVIDIA Jetson is a series of embedded computing boards " \
               "for robotics and AI at the edge, like the Jetson Nano and Jetson Orin."
    elif "ai" in q or "artificial intelligence" in q:
        return "AI (Artificial Intelligence) lets computers learn from data " \
               "and perform tasks that normally need human intelligence."
    elif "name" in q:
        return f"My name is {BOT_NAME}, your friendly QA bot!"
    elif "bye" in q or "exit" in q or "quit" in q:
        return "bye"
    else:
        return "Sorry, I don't understand that. " \
               "Try asking about: hello, python, jetson, ai, or my name."

def run_qa_bot():
    """QA Bot mode."""
    print(f"\n=== {BOT_NAME} QA Bot ===")
    print("Type 'bye' to exit.\n")
    while True:
        question = input("You: ")
        response = qa_answer(question)
        if response == "bye":
            print(f"{BOT_NAME}: Goodbye! Have a great day!")
            break
        print(f"{BOT_NAME}: {response}")
```
